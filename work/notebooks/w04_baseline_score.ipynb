## 1. My rule and its reason codes

## Baseline Rule

Pages with high impressions, low CTR, and an average search position between 5 and 20 should be prioritized because they already receive visibility but have the greatest opportunity to gain additional clicks through SEO improvements.

### Reason Codes

- CONFIRMED – High impressions, low CTR, and position between 5–20.
- MIXED – Meets some conditions but not all.
- OPPOSITE – Low impressions or already ranks very well.
- FALSE – Insufficient data or does not meet the baseline rule.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import os
import glob

csv_files = glob.glob("data/raw/*.csv")
print(csv_files)

df = pd.read_csv(csv_files[0])
df.head()
```

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

## Ranked Queue

The baseline score is based on impressions, average position, and CTR.

Pages with high impressions, low CTR, and an average position between 5 and 20 receive the highest priority. Lower scores are assigned to pages with poor visibility or already excellent rankings.

The ranked queue is exported as `work/outputs/baseline_action_score.csv`.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os

# Create output folder
os.makedirs("work/outputs", exist_ok=True)

# Baseline scoring rule
df["baseline_score"] = (
    (df["impressions_90d"] / df["impressions_90d"].max()) * 0.5
    + ((20 - df["avg_position"]).clip(lower=0) / 20) * 0.3
    + ((100 - df["ctr"]) / 100) * 0.2
)

# Highest score first
ranked = df.sort_values("baseline_score", ascending=False)

# Save CSV
ranked.to_csv(
    "work/outputs/baseline_action_score.csv",
    index=False
)

print("Saved successfully!")
ranked.head(10)
```
